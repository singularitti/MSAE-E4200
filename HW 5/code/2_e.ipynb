```julia
sq3 = 1/sqrt(3);
p = exp(2im*pi/3);
q = exp(4im*pi/3);
```

```julia
v1 = sq3 * [1 1 1];
v2 = sq3 * [1 p q];
v3 = sq3 * [1 q p];
```

```julia
U = vcat(v1, v2, v3)
```

```julia
transpose(conj(U)) * kappa * U
```
